# Step 5: The agent
A ReAct loop with native tool calling and seven tools: search_filings, get_filing_list, compare_sections, extract_metrics, risk_language_diff (new, dropped and re-toned risk sentences between two years), survey_banks (each bank searched separately for questions across banks) and calculate (safe arithmetic, so computed figures show their inputs). The user's role is applied inside every tool; every step is logged.

```python
import sys, json, pandas as pd
sys.path.insert(0, '../src')
pd.set_option('display.max_colwidth', 120)
```

```python
from jarvis.agent import Agent
ag = Agent()
QS = ["What new risks did RBC add in 2025 that were not in 2024?", "How does TD's credit risk language compare to BMO's in their most recent annual reports?", "Which of the Big Five banks has the most detailed climate risk disclosure?", "How did Scotiabank's provision for credit losses change from fiscal 2024 to fiscal 2025?", "Which banks mention AI risk or OSFI E-23 in their most recent filings?"]
runs = [ag.run(q) for q in QS]
```

```python
pd.DataFrame([{'question': r['question'], 'iterations': r['iterations'], 'tools': [t['tool'] for t in r['tool_calls']], 'usd': r['usd'], 'seconds': r['seconds']} for r in runs])
```

```python
for r in runs:
    print('#', r['question'])
    print(r['answer'][:1500], '\n')
```
